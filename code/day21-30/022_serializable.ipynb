在 Python 中，我们可以将程序中的数据以 JSON 格式进行保存

{
    name: "骆昊",
    age: 40,
    friends: ["王大锤", "白元芳"],
    cars: [
        {"brand": "BMW", "max_speed": 240},
        {"brand": "Benz", "max_speed": 280},
        {"brand": "Audi", "max_speed": 280}
    ]
}

json模块有四个比较重要的函数，分别是：

* dump - 将 Python 对象按照 JSON 格式序列化到文件中
* dumps - 将 Python 对象处理成 JSON 格式的字符串
* load - 将文件中的 JSON 数据反序列化成对象
* loads - 将字符串的内容反序列化成 Python 对象

```python
import json

my_dict = {
    'name': '骆昊',
    'age': 40,
    'friends': ['王大锤', '白元芳'],
    'cars': [
        {'brand': 'BMW', 'max_speed': 240},
        {'brand': 'Audi', 'max_speed': 280},
        {'brand': 'Benz', 'max_speed': 280}
    ]
}
# 将字典转换为 JSON 字符串
# # 注意：json.dumps() 方法返回的是字符串，不是字典
print(json.dumps(my_dict))

# 序列化到文件中
with open('data.json', 'w') as file:
    json.dump(my_dict, file)

with open('data.json', 'r') as file:
    # 将 JSON 字符串转换为字典
    my_dict = json.load(file)
    print(type(my_dict))
    print(my_dict)
    
with open('data.json', 'r') as file:
    # 将 JSON 字符串转换为字典
    # loads  =  load str
    my_dict2 = json.loads(file.read())
    print(type(my_dict2))
    print(my_dict2)
```

# requests库

requests库是一个用于发送HTTP请求的Python库

```python
# conda install requests
import requests

response = requests.get('https://api.github.com/events')
if response.status_code == 200 or response.status_code == 403:
    # 将 JSON 字符串转换为 Python 对象
    data = response.json()
    print(type(data))
    print(data)
```
