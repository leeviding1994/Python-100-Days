# 正则表达式

## 可以分为几种类型

| 类别 | 作用   | 例子            |
| -- | ---- | ------------- |
| 字符 | 匹配什么 | `a` `.` `\d`  |
| 数量 | 匹配多少 | `* + ? {n,m}` |
| 位置 | 匹配哪里 | `^ $ \b`      |
| 逻辑 | 如何组合 | `()` `        |


## 字符

\d    数字

\w    字母+数字+_

\s    空白

.     任意字符（除换行）

# 数量

\*     0次或多次

\+     1次或多次

?     0或1次

{n}   n次

{n,}  至少n次

{n,m} n到m次

# 位置

^     开头

$     结尾

\b    单词边界 ： 匹配单词边界，即字母、数字或下划线的边界

\B    非单词边界： 匹配非单词边界，如空格、换行符等，即下一个字符是字母、数字或下划线时则匹配

# 逻辑

()    组合

[]    包含

\    转义

|    或
