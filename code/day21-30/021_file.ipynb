使用 Python 内置的**open**函数来打开文件

通过函数的参数指定文件名、操作模式和字符编码等信息

| 参数 | 描述 |
| --- | --- |
 `'r'`    | 读取 （默认）                    |
| `'w'`    | 写入（会先截断之前的内容）       |
| `'x'`    | 写入，如果文件已经存在会产生异常 |
| `'a'`    | 追加，将内容写入到已有文件的末尾 |
| `'b'`    | 二进制模式                       |
| `'t'`    | 文本模式（默认）                 |
| `'+'`    | 更新（既可以读又可以写）         |

```python
file = open('致橡树.txt', 'r', encoding='utf-8')
print(file.read())
file.close()
```

```python
# 也可以使用with语句来打开文件，这样就不需要手动关闭文件了
with open('致橡树.txt', 'r', encoding='utf-8') as file:
    # 使用for循环逐行读取文件
    for line in file:
        print(line.strip())

# 还可以使用readlines()方法来读取文件的每一行，返回一个列表
with open('致橡树.txt', 'r', encoding='utf-8') as file:
    lines = file.readlines()
    for line in lines:
        print(line.strip())  # 使用strip()去掉每行末尾的换行符 
```

可以在打开文件时使用w或者a作为操作模式

a模式表示追加写入，不会覆盖文件内容

w模式表示写入文件，会覆盖文件内容

```python
with open('致橡树.txt', 'a', encoding='utf-8') as file:
    file.write('\n标题：《致橡树》')
    file.write('\n作者：舒婷')
    file.write('\n时间：1977年3月')
```

异常处理

使用 try except finally 来处理异常


```python
file = None
try:
    # r+：读写模式，文件存在则打开，不存在则报错
    file = open('致橡树.txt', 'r+', encoding='utf-8')
    print(file.read())
except FileNotFoundError:
    print('无法打开指定的文件!')
except LookupError:
    print('指定了未知的编码!')
except UnicodeDecodeError:
    print('读取文件时解码错误!')
finally:
    if file:
        file.close()
```

Python 中所有的异常都是BaseException的子类型

有四个直接的子类，分别是：SystemExit、KeyboardInterrupt、GeneratorExit和Exception

* SystemExit：系统退出异常，用于表示程序正常退出
* KeyboardInterrupt：键盘中断异常，用于表示用户通过Ctrl+C中断程序
* GeneratorExit：生成器退出异常，用于表示生成器函数异常退出
* Exception：基类，用于表示其他异常



```python
# 可以使用raise关键字来引发异常（抛出异常对象），而调用者可以通过try...except...结构来捕获并处理异常

class InputError(ValueError):
    """自定义异常类型"""
    pass


def fac(num):
    """求阶乘"""
    if num < 0:
        raise InputError('只能计算非负整数的阶乘')
    if num in (0, 1):
        return 1
    return num * fac(num - 1)

flag = True
while flag:
    num = int(input('n = '))
    try:
        print(f'{num}! = {fac(num)}')
        flag = False
    except InputError as err:
        print(err)
```

对于open函数返回的文件对象，还可以使用with上下文管理器语法在文件操作完成后自动执行文件对象的close方法

并不是所有的对象都可以放在with上下文语法中，只有符合上下文管理器协议（有__enter__和__exit__魔术方法）的对象才能使用这种语法

Python 标准库中的contextlib模块也提供了对with上下文语法的支持，后面用到的时候再为大家进行讲解。

```python
with open('致橡树.txt', 'r', encoding='utf-8') as file:
    print(file.read())
```

# 读写二进制文件

二进制文件在使用open函数打开文件时，

如果要进行读操作，操作模式是'rb'，

如果要进行写操作，操作模式是'wb'

> 读写文本文件时，read方法的返回值以及write方法的参数是str对象（字符串），而读写二进制文件时，read方法的返回值以及write方法的参数是bytes-like对象（字节串）


Python 的作用域规则：

全局作用域：模块级别的变量

局部作用域：函数内的变量

嵌套作用域：嵌套函数内的变量

with 语句、for 循环、if 语句等代码块不会创建新的作用域

```python
with open('kenan.jpeg', 'rb') as file1:
    data = file1.read()
with open('kenan_copy.jpeg', 'wb') as file2:
    #  写入数据到文件，with不会创建作用域
    file2.write(data)
```

```python
# 如果文件较大，可以分块读取和写入
try:
    with open('kenan.jpeg', 'rb') as file1, open('kenan_copy.jpeg', 'wb') as file2:
        data = file1.read(512)
        while data:
            file2.write(data)
            data = file1.read()
except FileNotFoundError:
    print('指定的文件无法打开.')
except IOError:
    print('读写文件时出现错误.')
print('程序执行结束.')
```
