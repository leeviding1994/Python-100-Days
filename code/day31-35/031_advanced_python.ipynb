```python
# 嵌套列表的坑

names = ['关羽', '张飞', '赵云', '马超', '黄忠']
courses = ['语文', '数学', '英语']
# 录入五个学生三门课程的成绩

# 不能这样写，因为会将所有行的引用都指向同一个列表
# 因为Python的列表是可变对象，所以当我们将一个列表赋值给多个变量时，它们会指向同一个列表
# [None] * len(courses) 会将 None 复制 len(courses) 次，生成一个列表
# [[None] * len(courses)] * len(names) 会将 [None] * len(courses) 复制 len(names) 次，生成一个列表
# [None] * len(courses) 不会导致是因为 [None] 是不可变对象，所以不会指向同一个列表
# scores = [[None] * len(courses)] * len(names)

# 正确的写法
scores = [[None] * len(courses) for _ in range(len(names))]
for row, name in enumerate(names):
    for col, course in enumerate(courses):
        scores[row][col] = float(input(f'请输入{name}的{course}成绩: '))
        print(scores)
```

```python
# 对排序

import heapq

list1 = [34, 25, 12, 99, 87, 63, 58, 78, 88, 92]
print(heapq.nsmallest(3, list1))  # [12, 25, 34]
print(heapq.nlargest(3, list1))  # [99, 88, 78]

list2 = [
    {'name': 'IBM', 'shares': 100, 'price': 91.1},
    {'name': 'AAPL', 'shares': 50, 'price': 543.22},
    {'name': 'FB', 'shares': 200, 'price': 21.09},
    {'name': 'HPQ', 'shares': 35, 'price': 31.75},
    {'name': 'YHOO', 'shares': 45, 'price': 16.35},
    {'name': 'ACME', 'shares': 75, 'price': 115.65}
]
print(heapq.nlargest(2, list2, key=lambda x: x['price']))
print(heapq.nlargest(2, list2, key=lambda x: x['shares']))
```

```python
# 内置函数示例：判断与反射

import builtins

class Person:
    pass


class Student(Person):
    def __init__(self, name):
        self.name = name


stu = Student('骆昊')

print('all([True, 1, 3 > 2]) =', builtins.all([True, 1, 3 > 2]))
print('any([0, "", 5]) =', builtins.any([0, "", 5]))
print('isinstance(stu, Student) =', builtins.isinstance(stu, Student))
print('issubclass(Student, Person) =', builtins.issubclass(Student, Person))

print('type(stu) =', builtins.type(stu))
print('id(stu) =', builtins.id(stu))
print("'name' in dir(stu) =", 'name' in builtins.dir(stu))
print('hasattr(stu, "age") =', builtins.hasattr(stu, 'age'))
print('getattr(stu, "name") =', builtins.getattr(stu, 'name'))

builtins.setattr(stu, 'age', 18)
print('setattr 后 stu.age =', stu.age)
builtins.delattr(stu, 'age')
print('delattr 后 hasattr(stu, "age") =', builtins.hasattr(stu, 'age'))
```

面向对象设计原则

单一职责原则 （SRP）- 一个类只做该做的事情（类的设计要高内聚）
开闭原则 （OCP）- 软件实体应该对扩展开发对修改关闭
依赖倒转原则（DIP）- 面向抽象编程（在弱类型语言中已经被弱化）
里氏替换原则（LSP） - 任何时候可以用子类对象替换掉父类对象
接口隔离原则（ISP）- 接口要小而专不要大而全（Python中没有接口的概念）
合成聚合复用原则（CARP） - 优先使用强关联关系而不是继承关系复用代码
最少知识原则（迪米特法则，LoD）- 不要给没有必然联系的对象发消息

```python
# 迭代器

class Fib(object):
    """迭代器"""
    
    def __init__(self, num):
        self.num = num
        self.a, self.b = 0, 1
        self.idx = 0
   
    def __iter__(self):
        return self

    def __next__(self):
        if self.idx < self.num:
            self.a, self.b = self.b, self.a + self.b
            self.idx += 1
            return self.a
        raise StopIteration()

my_lib = Fib(10)
for _ in range(10):
    print(next(my_lib))


# 生成器
# yield 语句 用于返回一个值，但是不会结束函数的执行
# 而是将函数的执行状态保存起来，等待下一次调用 next() 方法时继续执行
def fib(num):
    """生成器"""
    a, b = 0, 1
    for _ in range(num):
        a, b = b, a + b
        yield a

# 这里会使用协程的 next() 方法，每次调用 next() 方法时，会从上次调用的地方继续执行，直到遇到 yield 语句
# 什么是协程？ 协程是一种特殊的函数，它可以暂停执行，等待外部事件发生，然后继续执行
# 协程是在一个函数内模拟异步操作，而不需要使用回调函数或线程
gen = fib(10)
for _ in range(10):
    print(next(gen))
```

```python
# 现代协程主要使用await
import asyncio

async def task():
    print("start")
    await asyncio.sleep(1)
    print("end")

# 在 Jupyter / VS Code Notebook 中，事件循环已经在运行，直接 await 即可
# await 会暂停当前协程的执行，等待 await 后面的表达式执行完毕，然后继续执行
# 这里使用await，其实相当于这块代码都是Jupyter Notebook的一个函数
await task()

# 如果把这段代码保存成普通 .py 文件运行，再使用 asyncio.run(task())
```

```python
# 线程创建

import glob
import os
import threading

from PIL import Image

PREFIX = 'thumbnails'


def generate_thumbnail(infile, size, format='PNG'):
	file, ext = os.path.splitext(infile)
	file = file[file.rfind('/') + 1:]
	outfile = f'{PREFIX}/{file}_{size[0]}_{size[1]}.{ext}'
	img = Image.open(infile)
	img.thumbnail(size, Image.Resampling.LANCZOS)
	img.save(outfile, format)


def main():
	if not os.path.exists(PREFIX):
		os.mkdir(PREFIX)
	for infile in glob.glob('images/*.png'):
		for size in (32, 64, 128):
            # 创建并启动线程
			threading.Thread(
				target=generate_thumbnail, 
				args=(infile, (size, size))
			).start()
			

if __name__ == '__main__':
	main()
```
