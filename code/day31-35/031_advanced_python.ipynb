```python
# 生成式

prices = {
    'AAPL': 191.88,
    'IBM': 149.24,
    'ORCL': 48.44,
    'ACN': 166.89,
    'GOOG': 1186.96,
    'FB': 208.09,
    'SYMC': 21.29
}

print(prices)
# 新的prices 过滤 + 排序
list1 = [key for key, value in prices.items() if value > 200]
print(sorted(list1, reverse=False))
dict1 = {key: value for key, value in prices.items() if value > 200}
print(dict1)
```

```python
# 嵌套列表的坑

names = ['关羽', '张飞', '赵云', '马超', '黄忠']
courses = ['语文', '数学', '英语']
# 录入五个学生三门课程的成绩

# 不能这样写，因为会将所有行的引用都指向同一个列表
# 因为Python的列表是可变对象，所以当我们将一个列表赋值给多个变量时，它们会指向同一个列表
# [None] * len(courses) 会将 None 复制 len(courses) 次，生成一个列表
# [[None] * len(courses)] * len(names) 会将 [None] * len(courses) 复制 len(names) 次，生成一个列表
# [None] * len(courses) 不会导致是因为 [None] 是不可变对象，所以不会指向同一个列表
# scores = [[None] * len(courses)] * len(names)

# 正确的写法
scores = [[None] * len(courses) for _ in range(len(names))]
for row, name in enumerate(names):
    for col, course in enumerate(courses):
        scores[row][col] = float(input(f'请输入{name}的{course}成绩: '))
        print(scores)
```

```python
# 对排序

import heapq

list1 = [34, 25, 12, 99, 87, 63, 58, 78, 88, 92]
print(heapq.nsmallest(3, list1))  # [12, 25, 34]
print(heapq.nlargest(3, list1))  # [99, 88, 78]

list2 = [
    {'name': 'IBM', 'shares': 100, 'price': 91.1},
    {'name': 'AAPL', 'shares': 50, 'price': 543.22},
    {'name': 'FB', 'shares': 200, 'price': 21.09},
    {'name': 'HPQ', 'shares': 35, 'price': 31.75},
    {'name': 'YHOO', 'shares': 45, 'price': 16.35},
    {'name': 'ACME', 'shares': 75, 'price': 115.65}
]
print(heapq.nlargest(2, list2, key=lambda x: x['price']))
print(heapq.nlargest(2, list2, key=lambda x: x['shares']))
```

```python
# iterator 模块
import itertools

# 产生ABCD的全排列
print(list(itertools.permutations('ABCD')))
# 产生ABCDE的五选三组合
print(list(itertools.combinations('ABCDE', 3)))
# 产生ABCD和123的笛卡尔积
print(list(itertools.product('ABCD', '123')))
# 产生ABC的无限循环序列
letters = itertools.cycle(('A', 'B', 'C'))
for _ in range(7):
    print(next(letters), end=' ')
```

```python
# collections模块

import collections

"""
找出序列中出现次数最多的元素
"""
from collections import Counter

words = [
    'look', 'into', 'my', 'eyes', 'look', 'into', 'my', 'eyes',
    'the', 'eyes', 'the', 'eyes', 'the', 'eyes', 'not', 'around',
    'the', 'eyes', "don't", 'look', 'around', 'the', 'eyes',
    'look', 'into', 'my', 'eyes', "you're", 'under'
]
counter = Counter(words)
#  找出前3个出现次数最多的元素
print(counter.most_common(3))
```
